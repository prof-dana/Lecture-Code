```python
df1 = pd.read_csv("https://raw.githubusercontent.com/prof-dana/Lecture-Code/refs/heads/main/BigDataMining/data/sales-jan-2014.csv")
df2 = pd.read_csv("https://raw.githubusercontent.com/prof-dana/Lecture-Code/refs/heads/main/BigDataMining/data/sales-feb-2014.csv")
print(df1.shape)
print(df2.shape)
```

```python
df_list = [df1, df2]
result = pd.concat(df_list, axis=0) # Concatenate vertically along the rows
result = result.reset_index(drop=True) # Reset the index and discard the old index
print(result.shape)
```

```python
result.head()
```

```python
rows = result.shape[0]

newCol = np.random.uniform(0,1, rows).reshape(-1,1)
newCol[:5]
```

```python
newCol = pd.DataFrame(newCol, columns=['Random'], index=result.index) #transform np array to DataFrame
result2 = pd.concat([result, newCol], axis=1)
result2.head()
```
