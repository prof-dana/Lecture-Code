```python
class Human:
    def __init__(self, age, name): #Constructor
        self.age = age
        self.name = name

    def intro(self):#Method (optional)
        print(f"{self.name}({self.age})")

kim = Human(29, "KBU")
lee = Human(45, "AI")
```
