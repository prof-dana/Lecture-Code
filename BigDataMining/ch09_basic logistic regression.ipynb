```python
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/prof-dana/Lecture-Code/refs/heads/main/BigDataMining/data/UVA_Internet_Survey_Dataset.csv")
df.head()
```

```python
cols = ['no', 'who', 'Country', 'Years on Internet']

df.drop(cols, axis = 1, inplace =True)
df.info()
```

```python
category_cols = ['Gender', 'Household Income',
                 'Sexual Preference','Education Attainment',
                 'Major Occupation', "Marital Status"]

for col in category_cols:
    df[col] = df[col].astype('category')#category datatype

df.dtypes
```

```python
df["Gender"]
```

```python
df_onehot = pd.get_dummies(df)  # One-hot encode categorical columns
df_onehot.shape
```

```python
df_onehot.isnull().sum()
```

```python
df_onehot.fillna({"Age": df_onehot['Age'].mean()}, inplace=True)
df_onehot.isnull().sum()
```

```python
y_data = df_onehot.pop("Newbie").values#numpy type
x_data = df_onehot.values

y_data.shape, x_data.shape
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
      x_data, y_data, test_size=0.33, random_state=42)

X_train.shape, X_test.shape
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# scaler.fit(X_train)               # Calculate the mean and standard deviation
# x_scaled = scaler.transform(X_train)  # Transform the data

x_scaled = scaler.fit_transform(X_train)
x_scaled[0]
```

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(penalty="l2", solver="lbfgs",
                        max_iter=1000,fit_intercept=True)

lr.fit(X_train, y_train)
```

```python
re_x = scaler.transform(X_test)  # Transform the data

lr.predict_proba(re_x)
```

```python
lr.predict(re_x)
```

```python
from sklearn.metrics import confusion_matrix

y_pred = lr.predict(re_x)

confusion_matrix(y_test, y_pred)
```

```python
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score

print("Accuracy:", accuracy_score(y_test, y_pred))     # Accuracy
print("Precision:", precision_score(y_test, y_pred))  # Precision
print("Recall:", recall_score(y_test, y_pred))        # Recall
print("F1 Score:", f1_score(y_test, y_pred))           # F1 score
```
