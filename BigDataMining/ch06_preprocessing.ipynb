```python
import pandas as pd
import numpy as np

df = pd.read_csv('http://raw.githubusercontent.com/prof-dana/Lecture-Code/refs/heads/main/BigDataMining/data/titanic.csv')
df.head(2).T
```

```python
# (1) Display floating-point values to two decimal places (optional)
pd.options.display.float_format = '{:.2f}'.format

# (2) Calculate the percentage of missing values in each column
df.isnull().sum() / len(df) * 100
```

```python
df.drop('Cabin', axis=1, inplace=True)
df.columns
```

```python
df.dropna(subset=["Survived"], inplace=True)
df.tail()
```

```python
df[df["Age"].notnull()].groupby(["Gender"])["Age"].mean()
```

```python
df[df["Age"].notnull()].groupby(["Pclass"])["Age"].mean()
```

```python
df.fillna({"Age":df.groupby("Pclass")["Age"].mean()}, inplace=True)
df.isnull().sum() / len(df) * 100
```

```python
ctg_columns = ["Pclass", "Gender", "Embarked"]

for c in ctg_columns:
    onehot = pd.get_dummies(df[c], prefix=c)#해당 열 원핫인코딩

    #index로 병합하여 다시 df에 저장 df = pd.concat([df, onehot], axis=1)
    df = pd.merge(df,onehot, how="inner", left_index=True, right_index=True)

df.head(2).T
```

```python
ctg_columns = ["Pclass", "Gender", "Embarked"]

for c in ctg_columns:
    onehot = pd.get_dummies(df[c], prefix=c)#one-hot

    df = pd.concat([df, onehot], axis=1)#기존 데이터와 열 합치기
    df.drop(c, axis=1, inplace=True)#원본 열은 삭제

df.head(2).T
```
