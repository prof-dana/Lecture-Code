```python
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/prof-dana/Lecture-Code/refs/heads/main/MachineLearning/data/UVA_Internet_Survey_Dataset.csv")
df.head()
```

```python
df.pop('no') #no열 제거
df.pop('who') #who열 제거
df.pop('Country') #Country열 제거
df.pop('Years on Internet')#Years on Internet열 제거

df.dtypes
```

```python
category_cols = ['Gender', 'Household Income', 
                 'Sexual Preference','Education Attainment', 
                 'Major Occupation', "Marital Status"]

for col in category_cols:
    df[col] = df[col].astype('category')#범주형 데이터 타입
 
df.dtypes
```

```python
df["Gender"]
```

```python
df_onehot = pd.get_dummies(df) #category타입만 변환
df_onehot.shape
```

```python
df_onehot.isnull().sum()
```

```python
df_onehot.fillna({"Age": df_onehot['Age'].mean()}, inplace=True)
df_onehot.isnull().sum()
```

```python
y_data = df_onehot.pop("Newbie").values
x_data = df_onehot.values

y_data.shape, x_data.shape
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

#x_scaled = scaler.fit(X)            # 평균과 표준편차 계산
#x_scaled = scaler.transform(X)   # 데이터 변환
x_scaled = scaler.fit_transform(x_data)
x_scaled[0]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
      x_scaled, y_data, test_size=0.33, random_state=42)

X_train.shape, X_test.shape
```

```python
from sklearn.linear_model import LogisticRegression

logreg = LogisticRegression(fit_intercept=True)
logreg.fit(X_train, y_train.flatten())
```

```python
re_x= scaler.transform(X_test)   # 데이터 변환

#넘파이 소수점 표현 세팅(생략가능)
np.set_printoptions(precision=4, suppress=True)

logreg.predict_proba(re_x)
```

```python
logreg.predict(re_x)
```

```python
from sklearn.metrics import confusion_matrix

y_true = y_test.copy()

y_pred = logreg.predict(re_x)

confusion_matrix(y_true, y_pred)
```

```python
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score

print("정확도:",accuracy_score(y_true, y_pred)) # 정확도
print("정밀도:",precision_score(y_true, y_pred)) # 정밀도
print("재현율:",recall_score(y_true, y_pred))   # 재현율
print("F1score:",f1_score(y_true, y_pred))      # F1 스코어
```
